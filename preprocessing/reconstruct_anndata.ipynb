# Load AnnData
RNA total count data was first extracted from Seurat in R and can now be reconstructed in Python using anndata, allowing for normalizations and filtering in the next processing step

```python
import anndata as ad
import pandas as pd
from scipy.io import mmread
from scipy.sparse import csr_matrix
import os
```

```python
data_dir = r"C:\Users\Nikla\Documents\MoBi_Master_FS.3\Internship AG Höfer\DynaVelo_preprocessing\data"
count_data = os.path.join(data_dir, "Exp4_ctr_rna_counts.mtx")
cell_data = os.path.join(data_dir, "Exp4_ctr_rna_cells.csv")
gene_data = os.path.join(data_dir, "Exp4_ctr_rna_genes.csv")
meta_data = os.path.join(data_dir, "Exp4_ctr_rna_metadata.csv")

output_dir = r"C:\Users\Nikla\Documents\MoBi_Master_FS.3\Internship AG Höfer\DynaVelo_testing\data"
```

```python
rna_X = csr_matrix(mmread(count_data).T)
rna_genes = pd.read_csv(gene_data, squeeze=True).values
rna_cells = pd.read_csv(cell_data, squeeze=True).values

obs = pd.read_csv(meta_data, index_col=0)

adata = ad.AnnData(X=rna_X, obs=obs)
adata.var_names = rna_genes
adata.obs_names = rna_cells

adata.write_h5ad(os.path.join(output_dir, "total_counts_raw.h5ad"))
```
